```python
import pandas as pd

results = pd.read_csv("../data/cleaned/2023_results_by_state_clean.csv")
register = pd.read_csv("../data/raw/2027_voter_register_by_state.csv")

print(set(results["state"]) - set(register["state"]))
print(set(register["state"]) - set(results["state"]))
```

```python
combined = results.merge(register, on="state", how="left", validate="one_to_one")
print(combined.shape)
combined[["state", "total_valid_votes", "registered_voters", "precision"]].head()
```

```python
combined.to_csv("../data/cleaned/states_combined.csv", index=False)
```
