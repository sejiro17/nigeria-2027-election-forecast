```python
import pandas as pd

results = pd.read_csv("../data/cleaned/2023_results_by_state_clean.csv")
register = pd.read_csv("../data/raw/2027_voter_register_by_state.csv")

print(set(results["state"]) - set(register["state"]))
print(set(register["state"]) - set(results["state"]))
```

```python
combined = results.merge(register, on="state", how="left", validate="one_to_one")
print(combined.shape)
combined[["state", "total_valid_votes", "registered_voters", "precision"]].head()
```

```python
combined["registered_voters"].isna().sum()
```

```python
combined.to_csv("../data/cleaned/states_combined.csv", index=False)
```

```python
zones = {
    "North Central": ["Benue", "Kogi", "Kwara", "Nasarawa", "Niger", "Plateau", "FCT"],
    "North East": ["Adamawa", "Bauchi", "Borno", "Gombe", "Taraba", "Yobe"],
    "North West": ["Jigawa", "Kaduna", "Kano", "Katsina", "Kebbi", "Sokoto", "Zamfara"],
    "South East": ["Abia", "Anambra", "Ebonyi", "Enugu", "Imo"],
    "South South": ["Akwa Ibom", "Bayelsa", "Cross River", "Delta", "Edo", "Rivers"],
    "South West": ["Ekiti", "Lagos", "Ogun", "Ondo", "Osun", "Oyo"],
}
state_to_zone = {s: z for z, states in zones.items() for s in states}
combined["zone"] = combined["state"].map(state_to_zone)

print(combined["zone"].isna().sum(), "states without a zone")
combined.groupby("zone")["registered_voters"].sum()
```

```python
combined.to_csv("../data/cleaned/states_combined.csv", index=False)
```
