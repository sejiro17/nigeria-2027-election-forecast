```python
import pandas as pd
df = pd.read_csv("../data/raw/2023_results_by_state_wikipedia.csv", header=[0, 1, 2])
df.head()
```

```python
df.columns = [
    " ".join(part for part in col if not part.startswith("Unnamed")).strip()
    for col in df.columns
]
df.columns.tolist()
```

```python
df.columns = [
    "state",
    "tinubu_votes", "tinubu_pct", "tinubu_25",
    "atiku_votes", "atiku_pct", "atiku_25",
    "obi_votes", "obi_pct", "obi_25",
    "kwankwaso_votes", "kwankwaso_pct", "kwankwaso_25",
    "others_votes", "others_pct",
    "total_valid_votes", "turnout_pct",
]
df.head()
```

```python
df["state"] = df["state"].str.replace(r"\[\d+\]", "", regex=True).str.strip()
df["state"].tolist()
```

```python
print(", ".join(df["state"]))
```

```python
df = df[df["state"].notna() & (df["state"] != "Total")].reset_index(drop=True)
print(len(df), "rows")
print(", ".join(df["state"]))
```

```python
df.dtypes
```

```python
num_cols = df.columns.drop("state")
for col in num_cols:
    df[col] = pd.to_numeric(
        df[col].str.replace("%", "").str.replace(",", "").str.strip(),
        errors="coerce",
    )
df.dtypes
```

```python
vote_cols = ["tinubu_votes", "atiku_votes", "obi_votes", "kwankwaso_votes", "others_votes"]
df["check_diff"] = df[vote_cols].sum(axis=1) - df["total_valid_votes"]
df[df["check_diff"] != 0][["state", "check_diff"]]
```

```python
df[df["state"].isin(["Kwara", "Yobe"])][["state"] + vote_cols + ["total_valid_votes"]]
```

```python
df.loc[df["state"] == "Yobe", "atiku_votes"] = 198567
df["check_diff"] = df[vote_cols].sum(axis=1) - df["total_valid_votes"]
df[df["check_diff"] != 0][["state", "check_diff"]]
```

```python
for c in ["tinubu", "atiku", "obi", "kwankwaso"]:
    df[f"{c}_pct_calc"] = (df[f"{c}_votes"] / df["total_valid_votes"] * 100).round(2)
    off = df[(df[f"{c}_pct_calc"] - df[f"{c}_pct"]).abs() > 0.05]
    print(c, "mismatches:", off["state"].tolist())
```

```python
helper_cols = [c for c in df.columns if c.endswith("_calc") or c == "check_diff"]
clean = df.drop(columns=helper_cols)
clean.to_csv("../data/cleaned/2023_results_by_state_clean.csv", index=False)
print(clean.shape)
```
