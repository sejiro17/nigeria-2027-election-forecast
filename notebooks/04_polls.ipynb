```python
import pandas as pd

poll = pd.read_csv("../data/raw/2027_poll_oxford_metro_by_zone.csv")
national = {"tinubu": 24469, "obi": 23402, "atiku": 18430}

for c, total in national.items():
    missing = poll[c].isna()
    if missing.sum() == 1:
        poll.loc[missing, c] = total - poll[c].sum()
        print(c, "gap filled")
    else:
        print(c, "has", missing.sum(), "gaps, can't fill")
poll
```

```python
for c in ["tinubu", "obi", "atiku"]:
    poll[f"{c}_pct"] = (poll[c] / poll["participants"] * 100).round(2)
poll.to_csv("../data/cleaned/2027_poll_by_zone_clean.csv", index=False)
poll
```
